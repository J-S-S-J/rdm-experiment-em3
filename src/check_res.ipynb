```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../results/michelle_ses1_20260507_123618.csv")
print(df)
```

```python
mean_acc = df.groupby('coherence', sort=True)['accuracy'].mean()

ax = mean_acc.plot(kind='bar', figsize=(6,4))
ax.set_xlabel('Coherence')
ax.set_ylabel('Mean accuracy')
ax.set_ylim(0, 1)
ax.set_title('Mean accuracy by coherence')
plt.tight_layout()
plt.show()
```

```python
mean_rt = df.groupby('coherence', sort=True)['reaction_time'].mean()

ax_rt = mean_rt.plot(kind='bar', figsize=(6,4))
ax_rt.set_xlabel('Coherence')
ax_rt.set_ylabel('Mean reaction time (s)')
ax_rt.set_title('Mean RT by coherence')
plt.tight_layout()
plt.show()
```
